```python
str1 = "A man, a plan, a canal: Panama"
str2 = ""


for i in str1:
    if ('A' <= i <= 'Z') or ('a' <= i <= 'z') or ('0' <= i <= '9'):
        str2 = str2 + i.lower()

print(str2)


rev = ""
for i in range(len(str2) - 1, -1, -1):
    rev = rev + str2[i]


if str2 == rev:
    print("Valid Palindrome")
else:
    print("Not a Palindrome")
```
