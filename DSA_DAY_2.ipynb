```python
#sum of even & odds number
n=int(input("Enter size:"))
print("Enter list element:")
arr=[]
even=0
odd=0

# Populate the list with user input
for i in range(n):
  ele=int(input("Enter element:"))
  arr.append(ele)

# Calculate sum of even and odd numbers
for num in arr:
  if num % 2 == 0:
    even=even+num
  else:
    odd=odd+num

print("Even sum, odd sum:")
print(f"Even sum: {even}, Odd sum: {odd}")
```

```python
#find the number is tech or not
# Tech Number Program in Python

num = int(input("Enter a number: "))

# Count digits
digits = len(str(num))

# Check if number of digits is even
if digits % 2 != 0:
    print(num, "is not a Tech Number.")
else:
    # Split the number into two equal parts
    half = digits // 2
    num_str = str(num)

    first_part = int(num_str[:half])
    second_part = int(num_str[half:])

    # Calculate square of the sum
    total = first_part + second_part
    square = total * total

    # Check if it is a Tech Number
    if square == num:
        print(num, "is a Tech Number.")
    else:
        print(num, "is not a Tech Number.")
```
