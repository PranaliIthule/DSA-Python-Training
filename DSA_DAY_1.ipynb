```python
no = int(input("Enter no:"))
sum_digits = 0

temp_no = no # Use a temporary variable to preserve the original number if needed

while temp_no > 0:
    digit = temp_no % 10  # Get the last digit
    sum_digits += digit   # Add the digit to the sum
    temp_no //= 10        # Remove the last digit

print(f"The sum of the digits of {no} is: {sum_digits}")
```
