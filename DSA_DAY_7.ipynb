```python
n = int(input())
price = list(map(int, input().split()))

span = []

for i in range(n):
    count = 1
    j = i - 1

    while j >= 0 and price[j] <= price[i]:
        count += 1
        j -= 1

    span.append(count)

for i in span:
    print(i, end=" ")
```

```python
n = int(input())
price = list(map(int, input().split()))

stack = []
span = []

for i in range(n):
    while len(stack) > 0 and price[stack[-1]] <= price[i]:
        stack.pop()

    if len(stack) == 0:
        span.append(i + 1)
    else:
        span.append(i - stack[-1])

    stack.append(i)

for i in span:
    print(i, end=" ")
```
