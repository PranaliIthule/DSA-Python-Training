```python
no = int(input("Enter no:"))
sum_digits = 0

temp_no = no # Use a temporary variable to preserve the original number if needed

while temp_no > 0:
    digit = temp_no % 10  # Get the last digit
    sum_digits += digit   # Add the digit to the sum
    temp_no //= 10        # Remove the last digit

print(f"The sum of the digits of {no} is: {sum_digits}")
```

```python
no=int (input("Enter number:"))
or_no = no
rev = 0
while no > 0:
  rem = no % 10
  rev = rev * 10 + rem
  no = no // 10

if or_no == rev:
  print(f"{or_no} is a palindrome number.")
else:
  print(f"{or_no} is not a palindrome number.")
```

```python

#check no is armstrong
no=int (input("Enter number:"))
sum=0
save=no
count=0
while no>0:
  no=no//10
  count=count+1
  save =no
while no>0:
  rem=no%10
  sum=sum+(rem**count)
  no=no//10
if sum==save:
    print("no is armstrong")
else:
      print("no is not armstrong")
```

```python
# The number is a strong number (or 'person number') if it equals the sum of the factorial of its digits.
no = int(input("Enter number:"))
original_no = no
sum_of_factorials = 0


temp_no = no
while temp_no > 0:
    rem = temp_no % 10
    factorial = 1

    for i in range(1, rem + 1):
        factorial *= i
    sum_of_factorials += factorial


if original_no == sum_of_factorials:
    print(f"{original_no} is a strong number.")
else:
    print(f"{original_no} is not a strong number.")
```
