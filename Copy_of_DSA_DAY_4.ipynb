```python
#bubbleSort ascending order
def bubbleSort(arr):
    n = len(arr)
    for i in range(n - 1):
        for j in range(0, n - i - 1):
            if arr[j] > arr[j+1]:
                arr[j], arr[j+1] = arr[j+1], arr[j] # Perform the swap

if __name__ == "__main__":
    arr = [6, 23, 2, 4, 1, 8, 56, 3]
    bubbleSort(arr)
    print("Sorted array:", arr)
```

```python
#selection short Array (ascending order)
def selectionSort(arr):
    n = len(arr)

    for i in range(n):

        loc = i
        for j in range(i + 1, n):
            if arr[j] < arr[loc]:
                loc = j

        arr[i], arr[loc] = arr[loc], arr[i]

if __name__ == "__main__":
    arr = [6, 23, 2, 4, 1, 8, 56, 3]
    selectionSort(arr)
    print("Sorted array:", arr)
```

```python
# Insertion Sort : Ascending Order

arr = [5,2,17,3,4,9,15,1]

for i in range(1, len(arr)):
    current = arr[i]
    pos = i

    while pos > 0 and current < arr[pos-1]:
        arr[pos] = arr[pos-1]
        pos = pos - 1

    arr[pos] = current

print(arr)
```

```python
class stacks:
    def __init__(self):
        self.stack = []
        self.top = -1
        self.CAPACITY = 100

    def isfull(self):
        return self.top == self.CAPACITY - 1

    def isempty(self):
        return self.top == -1

    def push(self, ele):
        if self.isfull():
            print("Stack is full")
        else:
            self.stack.append(ele)
            self.top += 1

    def pop(self):
        if self.isempty():
            print("Stack is empty")
        else:
            ele = self.stack.pop()
            self.top -= 1
            return ele

if __name__ == '__main__':
    obj = stacks()

    arr = [1, 2, 3, 4, 5]

    for i in range(len(arr)):
        obj.push(arr[i])


    rev = []
    for i in range(len(arr)):
        rev.append(obj.pop())

    print("Original:", arr)
    print("Reversed:", rev)
```
